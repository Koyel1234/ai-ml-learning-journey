````python
from langchain_community.document_loaders import TextLoader
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

model = ChatOpenAI()

prompt = PromptTemplate(
    template = 'Write a summary for the following. \n {poem}',
    input_variables = ['poem']
)

parser = StrOutputParser()

# this is LCEL of RunnableSequence
chain = prompt | model | parser

print(chain.invoke({'poem': docs[0].page_content}))
````

- Here we'll learn about `eager loading` vs `lazy loading`.
- Till now we saw eager loading only - loading all documents in memory together in advance.

**Problems with Eager Loading:**<br>
There are two problems with eager loading.
- If we will try to print all documents e.g. metadata only (for sake of simplicity), it will take lots of time to print first doc as here all documents will be loaded together first, the first doc will be printed.
- Also here we are loading all pdfs in RAM together. If lots of pdfs are there and total memory required for all pdfs is huge amount of GBs, always it is not possible to load everything at once. 

So lazy loader is here to avoid above two problems. Normal loaders are eager loader but lazy loader uses generator concept in python which does lazy loading - load the document when actually it needs to be loaded, not in advance along with all others.

Added timer code in two code pieces will show time difference.

````python
from datetime import datetime
datetime.now()
````

````python
from datetime import datetime

# keeping load statement here as post next cell run, `docs` variable will be empty as in next cell `docs` is a generator variable. 
docs = loader.load()

current_time = datetime.now()  

for document in docs:
    print(document.metadata)
    print('Time needed:', datetime.now() - current_time)
````

````python
# lazy loader (should take less time to print first doc)
docs = loader.lazy_load()

current_time = datetime.now()  

for document in docs:
    print(document.metadata)
    print('Time needed:', datetime.now() - current_time )
````

**Observation:**

We can see documents are taking more time in lazy loading vs eager loading. This is actually normal and expected with small datasets like 3 PDFs. Here's why:

With only 3 small PDFs:
- Eager loading: Loads all 3 PDFs efficiently in one batch → Fast
- Lazy loading: Creates generator (overhead) → Loads first PDF on demand → Still needs to parse → Has generator overhead
The lazy loading overhead can outweigh its benefits when:

Files are small (3 PDFs in your case)
- Total data is small
- Files are already cached by OS

Lazy loading's real advantages appear with:
- Large number of documents (100+ PDFs) - Eager would take too long before you get first
- Very large files (each PDF is 100+ MB) - Eager would use massive RAM
- Only need subset (load 1 out of 1000 docs) - No point loading all 1000
- Memory constraints - Can't load everything at once

Example where lazy wins:
- Lazy: Get first doc in 0.1 seconds
- Eager: Get first doc in 5 seconds (loads all 1000)

So for few small pdfs:
- Each loop iteration is reading the PDF from disk and parsing it - the actual I/O work happens during the loop!
- That's why lazy is slower during iteration in case of 3 small pdfs.

Lazy loading's advantage is elsewhere:
- First document appears faster (before all are loaded)
- Lower memory usage (doesn't load everything at once)
- Faster if you only need some documents (don't waste time loading unused ones)

To see lazy loading's advantage, measure the time to first document:

````python
current_time = datetime.now()     
docs = loader.lazy_load()
first_doc = next(docs)            # Time to get FIRST document
print('lazy loading - Time to first:', datetime.now() - current_time)

# vs

current_time = datetime.now()     
docs = loader.load()              # All loaded before first is accessible
first_doc = docs[0]
print('Eager loading - Time to first:', datetime.now() - current_time)
````

So, in my case with 3 small PDFs:

Eager = ~0.05s (loads all)
Lazy = ~0.05-0.08s (generates + loads first + overhead)
So eager appears faster because the overhead of lazy doesn't pay off with such a small dataset!

`USER_AGENT` ia a http header that identifies my browser/client to a web server. Otherwise web-server may think this client is a bot and rejects the page scrapping request. Also cia this `USER_AGENT`, web-server detects device type (Windows/Android) and shows content (e.g. price of flipkart items) based on it.

Without using `USER_AGENT` though above cell is not throwing error, but while running `print(docs[0].page_content)` it's showing `Flipkart reCAPTCHAAre you a human?Confirming..` - it's asking for captcha confirmation that this client is human. So `USER_AGENT` will be required.
