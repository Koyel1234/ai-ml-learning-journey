```python
model = ChatOpenAI(model = "gpt-4", temperature = 0, max_completion_tokens = 10)
result = model.invoke('Write 5 line poem on Cricket.')
print(result)
print()
print(result.content)
```

```python
model = ChatOpenAI(model = "gpt-4", temperature = 2) # this will take much time to run it as max_completion_tokens is not passed (but why?)
result = model.invoke('Write 5 line poem on Cricket.')
print(result)
print(result.content)
```

```python
model = ChatOpenAI(model = "gpt-4", temperature = 2, max_completion_tokens = 10)
result = model.invoke('Write 5 line poem on Cricket.')
print(result)
print(result.content)
```

- If we want to get same output from same input, then we have to keep temperature around 0. If it's exactly 0 then output ill be always same from same input.
- As much we will increase temperature, 
same input will start generating different outputs.

### HuggingFace - Through API

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
llm = HuggingFaceEndpoint(
    repo_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task = "text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
result = model.invoke('What is the capital of India?')
print(result.content)
```

### HuggingFace - Download Locally

```python
# change location to store model if want to change location
# import os 
# os.environ['HF_HOME'] = 'D:/huggingface_cache'
```

```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline

llm = HuggingFacePipeline.from_model_id(
    model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task = "text-generation",
    pipeline_kwargs = dict(
        temparateure = 0.5,
        max_new_tokens = 100
    )
)

model = ChatHuggingFace(llm=llm)
result = model.invoke('What is the capital of India?')
print(result.content)
```

## Embedding Models

### OpenAI

```python
# single document
from langchain_openai import OpenAIEmbeddings
embedding = OpenAIEmbeddings(
    model = 'text-embedding-3-large',
    dimensions = 32 # manually fixed; min, max values are mentioned in documentation
)
result = embedding.embed_query("Delhi is the capital of India.")
print(str(result))
```

- Less dimention incurs less cost.

```python
print(result)
```

```python
print(len(result))
```

```python
# multiple document
from langchain_openai import OpenAIEmbeddings
embedding = OpenAIEmbeddings(
    model = 'text-embedding-3-large',
    dimensions = 32
)

documents = ["Delhi is the capital of India.",
             "Kolkata is the capital of West Bengal.",
             "Paris is the capital of France."
            ]

result = embedding.embed_documents(documents)
print(str(result))
```

```python
print(len(result))
```

```python
print(len(result[0]), len(result[1]), len(result[2]))
```

### HuggingFace

```python
from langchain_huggingface import HuggingFaceEmbeddings
embedding = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')

text = "Delhi is the capital of India."
result = embedding.embed_query(text) # dimension is predefined - 384
print(str(result))
```

```python
documents = ["Delhi is the capital of India.",
             "Kolkata is the capital of West Bengal.",
             "Paris is the capital of France."
            ]

result = embedding.embed_documents(documents)
print(str(result))
```
