### TextLoader

```python
# whatever be the doc loader, ultimately all loaders make different types of docuemnts in a standadizer format 'document object'
from langchain_community.document_loaders import TextLoader
from langchain_openai import ChatOpwenAI
from langchain_core.output_parsers import StrOutputparser
from langchain_core.prompts import PromptTemplate

load_dotenv()

model = ChatOpenAI()

prompt = PromptTemplate(
    template = 'Write a summary for the following. \n {poem}',
    input_variables = ['poem']
)

parser = StrOutputParser()

loader = TextLoader('cricket.txt', encoding='utf-8') # generate a text in cricket.txt file and upload in same directory of this notebook, if in diff folder, mention full path here
docs = loader.load()

print(docs)
```

```python
print(type(docs)) # python list # python loads all documents as list of documents
print(len(docs))
```

```python
print(docs[0]) # should have - 'page_content', 'metadata'
```

```python
print(type(docs[0])) 
```

```python
print(docs[0].page_content)
```

```python
print(docs[0].metadata)
```

```python
chain = prompt | model | parser
print(chain.invoke({'poem': docs[0].page_content}))
```

### DirectoryLoader

```python
# keep three pdfs within one folder
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader

loader = DirectoryLoader(
    path = 'books',
    glob = '*.pdf',
    loader_cls=PyPDFLoader
)

docs = loader.load()
print(len(docs)) # sum of all pages of all pdfs
```

```python
print(docs[0].page_content)
```

```python
print(docs[0].metadata)
```

```python
print(docs[234].page_content) # write last page number to see
```

```python
print(docs[234].metadata) # write last page number to see
```

```python
# will take lots of time to print first doc (as here all will be loaded together first, the first doc will be loaded)
for document in docs:
    print(document.metadata)
```

```python
# lazy loader (will take less time to print first doc)
docs = loader.lazy_load()
for document in docs:
    print(document.metadata)
```
