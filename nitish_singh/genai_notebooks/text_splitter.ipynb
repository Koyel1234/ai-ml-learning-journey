### Length Based Text Splitter

```python
from langchain.text_splitter import CharacterTextSplitter

text = """Write your own text"""

splitter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 0,
    separator = ''
)

result = splitter.split_text(text)
print(result) # list of chunks
```

```python
# doc loader + text split
from langchian_community.document_loaders import PyPDFLoader
from langchain.text_splitter import CharacterTextSplitter

loader = PyPDFLoader('mypdf.pdf') # load a pdf named mypdf.pdf
docs = loader.load() # one doc object for each page of pdf

splitter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 0,
    separator = ''
)

result = splitter.split_documents(docs)
print(result) # list of chunks
```

### Semantic Meaning Based Chunking

```python
text = """Two lines about topic1. Two lines about topic2.

Four lines about topic3.
"""

from langchian_experimental.text_splitter import SemanticChunker
from langchain_openai.embeddings import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

text_splitter = SemanticChunker(
    OpenAIEmbeddings(), 
    breakpoint_threshold_type = "standard_deviation", # there are different criteria, one of them is sd
    breakpoint_threshold_amount = 1 # try with different threshold
)

chunks = text_splitter.split_text(docs)
print(len(chunks))
print(chunks)
```
