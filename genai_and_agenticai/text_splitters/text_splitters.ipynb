### Length Based Text Splitter

Paste link of website to see chunk wise splitting.

- First let's see a simple example of text splitting example on small paragraph.

````python
from langchain_text_splitters import CharacterTextSplitter

text = """
Space exploration has led to incredible scientific discoveries. From landing on the Moon to exploring Mars, humanity continues to push the boundaries of what's possible beyond our plannet.

These missions have not only expanded our knowledge of the universe but have also contributed to advancements in technology here on Earth. Satelite communications, GPS, and even certain medical imaging techniques trace their roots back to innovations driven by space programs.
"""

splitter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 0,
    separator = ''
)

result = splitter.split_text(text)
print(result)
````

````python
print(len(result))
````

- Next see character wise text spilitting on a pdf file.
- Here both document loader and text splitter concepts are demonstrated.
- Here I have used `ai-ml-curriculum.pdf`, the same pdf used for learning PYPDFLoader concept in document_loaders learning part.

````python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import CharacterTextSplitter

loader = PyPDFLoader('ai-ml-curriculum.pdf') 

# one doc object for each page of pdf
docs = loader.load() 

splitter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 0,
    separator = ''
)

result = splitter.split_documents(docs)
````

````python
# list of chunks
print(result) 
````

````python
# total chunks got created out of 2 documents as it was having 1 page in the document
print(len(result))
````

````python
print(result[0])
````

````python
print(result[16])
````

````python
print(result[0].page_content) 
````

````python
print(result[0].metadata)
````

Here we have an important parameter `chunk_overlap` which prevents abrupt text cut in-between from a text or document. But we need trade-off here as has both advantages and disadvantages are there for this parameter. Generally 10-20 is good

### RecursiveChacracterTextSplitter

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """
Space exploration has led to incredible scientific discoveries. From landing on the Moon to exploring Mars, humanity continues to push the boundaries of what's possible beyond our plannet.

These missions have not only expanded our knowledge of the universe but have also contributed to advancements in technology here on Earth. Satelite communications, GPS, and even certain medical imaging techniques trace their roots back to innovations driven by space programs.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 0
)

chunks = splitter.split_text(text)
````

### Document-Structure Based Text Splitter

- First let's see example with Python code.

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

text = """
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade # grade is a float (like 8.5 or 9.2)

    def get_details(self):
        return self.name

    def is_passing(self):
        return self.grade >= 6.0

    

# Example usage
student1 = Student("Aarav", 20, 8.2)
print(student1.get_details())

if student1.is_passing():
    print("The student is passing.")
else:
    print("The student is not passing.")

"""

splitter = RecursiveCharacterTextSplitter.from_language(
    language = Language.PYTHON, 
    chunk_size = 300,
    chunk_overlap = 0
)

chunks = splitter.split_text(text)
````

````python
print(len(chunks))
````

````python
print(chunks)
````

````python
print(chunks[1])
````

- Now let's see example with Markdown code.

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

text = """
# Project Name: Smart Student Tracker

A simple Python-based project to manage and track student data, including their grades, age, and academic status.


## Features

- Add new students with relevant info
- View student details
- Check if a student is passing
- Easily extendable class-based design


## 🛠 Tech Stack

- Python 3.10+
- No external dependencies


## Getting Started

1. Clone the repo  
   ```bash
   git clone https://github.com/your-username/student-tracker.git

"""

splitter = RecursiveCharacterTextSplitter.from_language(
    language = Language.MARKDOWN, 
    chunk_size = 300,
    chunk_overlap = 0
)

chunks = splitter.split_text(text)
````

### Semantic Meaning Based Chunking

````python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai.embeddings import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

text_splitter = SemanticChunker(
    OpenAIEmbeddings(), 
    breakpoint_threshold_type = "standard_deviation", 
    breakpoint_threshold_amount = 1 
)

sample = """
Farmers were working hard in the fields, preparing the soil and planting seeds for the next season. The sun was bright, and the air smelled of earth and fresh grass. The Indian Premier League (IPL) is the biggest cricket league in the world. People all over the world watch the matches and cheer for their favourite teams.


Terrorism is a big danger to peace and safety. It causes harm to people and creates fear in cities and villages. When such attacks happen, they leave behind pain and sadness. To fight terrorism, we need strong laws, alert security forces, and support from people who care about peace and safety.
"""


chunks = text_splitter.split_text(sample)
````

````python
print(len(chunks))
````

````python
print(chunks)
````

````python
print(chunks[0])
````

````python
print(chunks[1])
````

````python
print(chunks[2])
````

There are different criteria to do splitting in semantic way, one of them is standard deviation.

This chunking strategy is very new, not in very much use in industry as of now, but it has much potential due to it's semantic nature.
