```python
from langchain_openai import OpenAIEmbeddings
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
embedding = OpenAIEmbeddings(
    model = 'text-embedding-3-large',
    dimensions = 32
)
```

```python
documents = [  
    "Virat Kohli is an Indian cricketer known for his aggressive batting and leadership.",
    "MS Dhoni is a former Indian captain famous for his calm demeanor and finishing skills.",
    "Sachin Tendulkar, also known as the 'God of Cricket', holds many batting records.",
    "Rohit Sharma is known for his elegant batting and record-breaking double centuries.",
    "Jasprit Bumrah is an Indian fast bowler known for his unorthodox action and yorkers."
]
```

```python
query = "Tell something about Bumrah."
```

```python
doc_embeddings = embedding.embed_documents(documents)
query_embedding = embedding.embed_query(query)
```

```python
scores = cosine_similarity([query_embedding], doc_embeddings)[0] # to get list
```

```python
scores
```

```python
list(enumerate(scores)) # gives scores with index
```

```python
index, score = sorted(list(enumerate(scores)), key=lambda x: x[1])[-1] # to keep positions fixed
```

```python
print(query)
print(documents[index])
print('similarity score: ', score)
```

```python
query = "Tell something about Paris."

doc_embeddings = embedding.embed_documents(documents)
query_embedding = embedding.embed_query(query)

scores = cosine_similarity([query_embedding], doc_embeddings)[0]
index, score = sorted(list(enumerate(scores)), key=lambda x: x[1])[-1] # to keep positions fixed

print(query)
print(documents[index])
print('all similarity score: ', scores)
print('max similarity score: ', score)
```

- Whatever vectors we are creating from documents, they are loaded into vector store/database (will see in later parts)
- This whole process of generating vectors of query and then getting similarity score of query vector with other vectors, taking the vector having max similarity with query vector as output is called retrival process.
