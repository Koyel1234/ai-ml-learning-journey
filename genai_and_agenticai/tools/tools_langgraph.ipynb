- Here we will learn how to integrate tools in LangGraph.
- To use tools, we have to create a tool_node as we do for chat nodes.

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Annotated
from langchain_core.messages import BaseMessage, HumanMessage
from langchain_openai import ChatOpenAI
from langgraph.graph.message import add_messages
from langgraph.prebuilt import ToolNode, tools_condition
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.tools import tool
import requests
import random
import os
```

```python
# calling LLM with default model 'gpt-3.5-turbo'
llm = ChatOpenAI()
```

- Below we will generate three tools: one built-in and other two custom tools.

```python
# Built-in tool
search_tool = DuckDuckGoSearchRun(region="us-en")



# custom tool
@tool
def calculator(first_num: float, second_num: float, operation: str) -> dict:
    """
    Perform a basic arithmetic operation on two numbers.
    Supported operations: add, sub, mul, div
    """
    try:
        if operation == "add":
            result = first_num + second_num
        elif operation == "subtract":
            result = first_num - second_num
        elif operation == "mul":
            result = first_num * second_num
        elif operation == "div":
            if second_num == 0:
                raise {"error": "Division by zero is not allowed."}
            result = first_num / second_num
        else:
            raise {"error": f"Unsupported operation: {operation}"}
        return {"first_num": first_num, "second_num": second_num, "operation": operation, "result": result}
    except Exception as e:
        raise {"error": str(e)}



STOCKPRICE_API_KEY = os.getenv("STOCKPRICE_API_KEY")

@tool
def get_stock_price(symbol: str) -> dict:
    """
    Fetch latest stock price for a given symbol )e.g. - 'AAPL', 'TSLA')
    using Alpha Vantage API key in the URL.
    """
    url = f"https://www.alphavantage.co/query?function=GLOBAL_QUOTE&symbol={symbol}&apikey={STOCKPRICE_API_KEY}" 
    response = requests.get(url) 
    return response.json() 
```

- In tool `get_stock_price` alphavantage api key is loaded in `STOCKPRICE_API_KEY` variable from .env file.
- This tool returns the JSON response from the API fot stock price.

```python
# Make tool list
tools = [get_stock_price, calculator, search_tool]

# make the LLM tool-aware
llm_with_tools = llm.bind_tools(tools)
```

```python
# state
class ChatState(TypedDict):
    messages: Annotated[list[BaseMessage], add_messages]
```

```python
# graph nodes
def chat_node(state: ChatState):
    messages = state["messages"]

    # send to model
    response = llm_with_tools.invoke(messages)
    return {"messages": [response]}

tool_node = ToolNode(tools) # Executes tool calls
```

```python
# graph structure
graph = StateGraph(ChatState)
graph.add_node('chat_node', chat_node)
graph.add_node('tools', tool_node)
```

- Above is showing graph's location in memory.

```python
graph.add_edge(START, 'chat_node')

# If the LLM asked for a tool, go to ToolNode; else finish
graph.add_conditional_edges('chat_node', tools_condition)

chatbot = graph.compile()
chatbot
```

```python
# regular chat
out = chatbot.invoke({"messages": [HumanMessage(content="Hello")]})
print(out["messages"][-1].content)
```

```python
# chat required tool `calculator`
out = chatbot.invoke({"messages": [HumanMessage(content="What is 2*3?")]})
print(out["messages"][-1].content)
```

```python
# chat required tool `get_stock_price`
out = chatbot.invoke({"messages": [HumanMessage(content="What is the stock price of apple?")]})
print(out["messages"][-1].content)
```

```python
# chat required both tools `get_stock_price` and `calculator`
out = chatbot.invoke({"messages": [HumanMessage(content="What is the price of 50 shares apple?")]})
print(out["messages"][-1].content)
```

- Here we are seeing that if user query needs tool call, then we are getting answer in json format, but that is not user friendly.
- In case tool is called to fetch or generate output, post that workflow should return answer in user friendly language.
- So, we have to make a little modification in the flow, from tool node it should again go to chat node to generate user friendly answer.

```python
# change the graph structure
graph = StateGraph(ChatState)
graph.add_node('chat_node', chat_node)
graph.add_node('tools', tool_node)

graph.add_edge(START, 'chat_node')

# If the LLM asked for a tool, go to ToolNode; else finish
graph.add_conditional_edges('chat_node', tools_condition)

# if going to `tools` node, return to `chat_node` from there
graph.add_edge("tools", "chat_node")

chatbot = graph.compile()
chatbot
```

```python
# regular chat - no tool is required
out = chatbot.invoke({"messages": [HumanMessage(content="Hello")]})
print(out["messages"][-1].content)
```

```python
# chat required tool `calculator`
out = chatbot.invoke({"messages": [HumanMessage(content="What is 2*3?")]})
print(out["messages"][-1].content)
```

```python
# chat required tool `get_stock_price`
out = chatbot.invoke({"messages": [HumanMessage(content="What is the stock price of apple?")]})
print(out["messages"][-1].content)
```

```python
# chat required both tools `get_stock_price` and `calculator`
out = chatbot.invoke({"messages": [HumanMessage(content="What is the price of 50 shares apple?")]})
print(out["messages"][-1].content)
```
