```python
x = "an apple a day keeps the apple a aa aaa aaaa aaaaa"
s = 'apple'



def check(x, s):
    i = 0 # x
    j = 0  # s
    c = 0
    while i< len(x) - len(s):
        if x[i] == s[j]:
            print(x[i], s[j],i, j)
            j += 1
            
            
        if j>0 and x[i] != s[j] and j < len(s)-1:
            j = 0   
            
        i += 1
        
        print(i)
    
        if j == len(s):
            c += 1
        # i =3, j=0  
    return c
    
check(x,s)
```
